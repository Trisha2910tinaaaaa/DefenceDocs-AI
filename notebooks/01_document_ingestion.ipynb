```python
import glob

pdf_files = glob.glob("/content/*.pdf")

print("Found:", len(pdf_files))

for pdf in pdf_files:
    print(pdf)
```

```python
from src.chunker import chunk_documents

chunks = chunk_documents(documents)

print("Total chunks:", len(chunks))
print(chunks[0])
```

```python
with open(
    "data/processed/chunks.json",
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        chunks,
        f,
        ensure_ascii=False,
        indent=2
    )
```

```python
from src.embeddings import create_embeddings

model, embeddings = create_embeddings(chunks)

print(embeddings.shape)
```

```python
from src.vector_store import create_index

index = create_index(embeddings)

print("Vectors stored:", index.ntotal)
```

```python
import sys

if "src.vector_store" in sys.modules:
    del sys.modules["src.vector_store"]
```

```python
from src.vector_store import create_index, search
```

```python
index = create_index(embeddings)

print("Vectors stored:", index.ntotal)
```
